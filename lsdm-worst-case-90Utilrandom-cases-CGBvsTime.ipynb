```python
import matplotlib.pyplot as plt
#import numpy as np
import pandas as pd
import math


names_bw=['time', 'bw', 'op1', 'write_size', 'op2' ]
names_latency=['time_msec', 'latency', 'op2', 'write_size', 'op3']
```

### Uniform Random on HM-LS

```python
df_90_uni = pd.read_csv('/home/surbhi/measurements/worst_case/lsdm/90Util/1MB/8TB/uniform/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_90_uni['time_seconds'] = df_90_uni['time_msec']/1000

# Round down the time_seconds to the nearest second
df_90_uni['time_seconds'] = df_90_uni['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_90_uni = df_90_uni.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_90_uni.columns = ['time_seconds', 'total_writes']
new_df_90_uni['total_writes'] = new_df_90_uni['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_90_uni.shape)


```

```python
#Create the line plots
plt.figure(figsize=(8, 6), dpi=80)
plt.rc('text', usetex=True)
plt.rc('font', family='serif')
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42


fig, ax = plt.subplots()
ax.plot(new_df_90_uni['time_seconds'], new_df_90_uni['total_writes'], linestyle='-')

# Add labels and title
plt.xlabel('Time taken in seconds', fontsize=16)
plt.xticks(fontsize=14)
plt.ylabel('Cumulative GBs written', fontsize=16)
plt.yticks(fontsize=14)
#avg_bw = df_bw['bw'].mean()
#avg_bw_str = str(round(avg_bw, 2)) + 'KB/sec'
#plt.axhline(y=avg_bw, color='r', linestyle='-', label='Avg Bandwidth: ' + avg_bw_str)
plt.title('LSDM, 5GB uniform-random\nIO Pattern (FG GC) (bs=1MB)(90\% Util)', fontsize=16)
labels = plt.gca().get_xticklabels()
label_to_change = labels[-1].set_color('red')
plt.legend(loc='upper left', fontsize=14)

# Show the plot
plt.grid(True)  # Optionally, add grid lines
plt.savefig('/tmp/tmp.pdf', bbox_inches='tight')
plt.show()
```

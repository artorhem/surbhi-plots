```python
import matplotlib.pyplot as plt
#import numpy as np
import pandas as pd
import math


names_bw=['time', 'bw', 'op1', 'write_size', 'op2' ]
```

### Uniform Random on HM-LS

```python
df_bw = pd.read_csv('/home/surbhi/measurements/worst_case/lsdm/90Util/1MB/8TB/uniform/bw_log_avg_bw.1.log', names=names_bw)
#process the bw data
nb_rows = 60
df_binned = df_bw.groupby(df_bw.index//nb_rows).mean()
# Calculate cumulative GBs of data written
df_binned['cumulative_gb'] = (df_binned['write_size'] * 60).cumsum() / (1024**3)
print(df_bw.shape)
print(df_binned.shape)
```

```python
#Create the line plots
plt.figure(figsize=(8, 6), dpi=80)
plt.rc('text', usetex=True)
plt.rc('font', family='serif')
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42


fig, ax = plt.subplots()
ax.plot(df_binned['cumulative_gb'], df_binned['bw'], linestyle='-')

# Add labels and title
plt.xlabel('Cumulative GBs Written (random writes to a disk)', fontsize=16)
plt.xticks(fontsize=14)
plt.ylabel('Bandwidth (KB/sec)', fontsize=16)
plt.yticks(fontsize=14)
avg_bw = df_bw['bw'].mean()
avg_bw_str = str(round(avg_bw, 2)) + 'KB/sec'
plt.axhline(y=avg_bw, color='r', linestyle='-', label='Avg Bandwidth: ' + avg_bw_str)
plt.title('HM-LS, bs=1MB, uniform-random\nIO with FG GC (90\% utilization)', fontsize=16)
labels = plt.gca().get_xticklabels()
label_to_change = labels[-1].set_color('red')
plt.legend(loc='upper left', fontsize=14)

# Show the plot
plt.grid(True)  # Optionally, add grid lines
plt.savefig('/tmp/tmp.pdf', bbox_inches='tight')
plt.show()
```

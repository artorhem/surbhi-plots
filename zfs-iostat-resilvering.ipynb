```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def convert_bandwidth_to_latency_df(csv_file):
    """
    Read CSV file with per-minute bandwidth data and convert to latency per MB
    Each bandwidth reading represents the bandwidth for that entire minute,
    so we expand it to 60 data points (one per second within that minute)
    
    Args:
        csv_file: Path to CSV file with bandwidth data (MB/s recorded per minute)
    
    Returns:
        DataFrame with expanded data and 'duration' column (latency in milliseconds per MB)
    """
    # Read the CSV file
    df = pd.read_csv(csv_file, names=['writebw'])
    
    # Clean and convert bandwidth data
    df['writebw'] = pd.to_numeric(df['writebw'].str.replace('M', ''), errors='coerce')
    
    # Remove any NaN values
    df = df.dropna()
    
    # Expand each minute's bandwidth to 60 seconds
    expanded_data = []
    for _, row in df.iterrows():
        # Add the same bandwidth value 60 times (once for each second in the minute)
        expanded_data.extend([row['writebw']] * 60)
    
    # Create new DataFrame with expanded data
    expanded_df = pd.DataFrame({'writebw': expanded_data})
    
    # Convert bandwidth to latency per MB (milliseconds)
    # Formula: Latency per MB (ms) = 1000 / Bandwidth (MB/s)
    expanded_df['duration'] = (1000 / expanded_df['writebw']).round(4)
    
    return expanded_df

# Create a more compact figure with reduced whitespace
plt.figure(figsize=(5, 4))

# Set up clean styling
sns.set_theme()
sns.set_style("white")
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.spines.left'] = True
plt.rcParams['axes.spines.bottom'] = True

# Colorblind-friendly palette (Paul Tol's vibrant palette)
colors = {
    'cmr': '#EE7733',      # Orange
    'dm_smr': '#0077BB',   # Blue  
    'ha_smr': '#33BBEE',   # Light blue
    'hm_hybrid': '#009988', # Teal
    'hm_ls': '#CC3311'     # Red
}

# Performance optimizations
plt.rcParams['path.simplify'] = True
plt.rcParams['path.simplify_threshold'] = 0.1

# File paths - update these to your actual file paths
file_paths = {
    'cmr': '/home/surbhi/measurements/zfs-resilvering-wb/writebw-cmr.csv',
    'dm_smr': '/home/surbhi/measurements/zfs-resilvering-status/DM-SMR/dm-smr-resilvering-bw.csv',
    'ha_smr': '/home/surbhi/measurements/zfs-resilvering-status/HA-SMR/ha-smr-resilver-bw.csv',
    'hm_hybrid': '/home/surbhi/measurements/zfs-resilvering-status/HM-Hybrid/hmhybrid-resilver.csv',
    'hm_ls': '/home/surbhi/measurements/zfs-resilvering-status/HM-LS/hmls-resilver.csv'
}

# Labels and line styles for each dataset
plot_configs = {
    'cmr': {'label': 'CMR', 'linestyle': '-', 'linewidth': 2.5},
    'dm_smr': {'label': 'DM-SMR', 'linestyle': '--', 'linewidth': 2.5},
    'ha_smr': {'label': 'HA-SMR', 'linestyle': '-.', 'linewidth': 2.5},
    'hm_hybrid': {'label': 'HM-Hybrid', 'linestyle': ':', 'linewidth': 3.0},
    'hm_ls': {'label': 'HM-LS', 'linestyle': ':', 'linewidth': 3.0}
}

# Plot CDF for each dataset
for key, file_path in file_paths.items():
    try:
        # Convert bandwidth to latency with proper minute-to-second expansion
        df = convert_bandwidth_to_latency_df(file_path)
        
        # Sort data for CDF calculation
        data_sorted = np.sort(df['duration'])
        
        # Calculate CDF
        cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
        
        # Plot the CDF
        config = plot_configs[key]
        plt.plot(data_sorted, cdf, 
                linestyle=config['linestyle'], 
                label=config['label'], 
                color=colors[key], 
                linewidth=config['linewidth'], 
                alpha=0.85)
                
        print(f"Processed {key}: {len(data_sorted)} data points")
        
    except FileNotFoundError:
        print(f"Warning: Could not find file {file_path}")
    except Exception as e:
        print(f"Error processing {key}: {str(e)}")

# Set log scale for x-axis
plt.xscale('log', base=2)

# Clean styling - no grid
plt.grid(False)

# Add percentile reference lines (only for 25th, 50th, 75th percentiles)
percentiles = [0.25, 0.5, 0.75]
for p in percentiles:
    plt.axhline(y=p, color='gray', linestyle='--', alpha=0.4, linewidth=1, zorder=0)

# Set ticks and labels with better spacing
ytick_positions = [0, 0.25, 0.5, 0.75, 1.0]
ytick_labels = ['0%', '25%', '50%', '75%', '100%']
plt.yticks(ytick_positions, ytick_labels, fontsize=11)
plt.xticks(fontsize=11)

# Labels and title with improved typography
plt.xlabel('Latency per MB (milliseconds)', fontsize=12, fontweight='medium')
plt.ylabel('CDF', fontsize=12, fontweight='medium')
plt.title('ZFS Resilvering Performance: CDF of Write Latency per MB)', 
          fontsize=13, fontweight='semibold', pad=15)

# Improved legend
plt.legend(fontsize=10, frameon=True, fancybox=False, 
           shadow=False, framealpha=0.9, edgecolor='lightgray',
           loc='lower right')

# Reduce whitespace around the plot
plt.tight_layout(pad=0.5)

# Remove top and right spines for cleaner look
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

# Save with high quality - increased DPI and sharpness
plt.savefig("ZFS_resilvering_CDF_corrected.png", bbox_inches='tight', dpi=600, 
            facecolor='white', edgecolor='none', format='png', 
            pil_kwargs={'optimize': True})

plt.show()

# Print some statistics for verification
with open("/home/surbhi/measurements/zfs-resilvering-status/zfs-resilvering-stats", "w") as f:
    print("\nDataset Statistics:", file=f)
    for key, file_path in file_paths.items():
        try:
            df = convert_bandwidth_to_latency_df(file_path)
            print(f"{plot_configs[key]['label']}:", file=f)
            print(f"  Total data points: {len(df)}")
            print(f"  Latency range: {df['duration'].min():.4f}ms to {df['duration'].max():.4f}ms", file=f)
            print(f"  Median latency: {df['duration'].median():.4f}ms", file=f)
            print(f"  25th percentile: {df['duration'].quantile(0.25):.4f}ms", file=f)
            print(f"  50th percentile: {df['duration'].quantile(0.50):.4f}ms", file=f)
            print(f"  90th percentile: {df['duration'].quantile(0.90):.4f}ms", file=f)
            print(f"  95th percentile: {df['duration'].quantile(0.95):.4f}ms", file=f)
            print(f"  99th percentile: {df['duration'].quantile(0.99):.4f}ms", file=f)
            print()
        except:
            print(f"{key}: Could not process file")
```

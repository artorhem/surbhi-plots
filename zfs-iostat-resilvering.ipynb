```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

def convert_bandwidth_to_latency_df(csv_file):
    """
    Read CSV file and convert bandwidth to latency per MB
    
    Args:
        csv_file: Path to CSV file
        bandwidth_column: Name of the column containing bandwidth data (MB/s)
    
    Returns:
        DataFrame with original data plus 'duration' column (latency in ms per MB)
    """
    # Read the CSV file
    df = pd.read_csv(csv_file, names=['writebw'])
    df['writebw'] = pd.to_numeric(df['writebw'].str.replace('M', ''), errors='coerce')
    
    # Convert bandwidth to latency per MB (ms)
    # Formula: Latency per MB (ms) = 1000 ms / Bandwidth (MB/s)
    df['duration'] = (1 / df['writebw']).round(2)
    
    # Round to 2 decimal places for readability
    df['duration'] = df['duration'].round(2)
    
    return df



# Create a more compact figure with reduced whitespace
plt.figure(figsize=(6, 4.5))

# Set up clean styling
sns.set_theme()
sns.set_style("white")
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.spines.left'] = True
plt.rcParams['axes.spines.bottom'] = True

# Colorblind-friendly palette (Paul Tol's vibrant palette)
colors = {
    '90%': '#EE7733',  # Orange
    '75%': '#0077BB',  # Blue  
    '50%': '#33BBEE',  # Light blue
    '25%': '#009988',  # Teal
    'extra': '#CC3311'  # Red - perfect for adding another series
}

# Performance optimizations
plt.rcParams['path.simplify'] = True
plt.rcParams['path.simplify_threshold'] = 0.1

#df1 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-cmr.csv', names=['writebw'])
df1 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-cmr.csv', names=['writebw'])
#df2 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-dmsmr.csv', names=['writebw'])
df2 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-status/DM-SMR/dm-smr-resilvering-bw.csv', names=['writebw'])
#df3 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-hasmr.csv', names=['writebw'])
df3 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-status/HA-SMR/ha-smr-resilver-bw.csv', names=['writebw'])
#df4 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-hmhybrid.csv', names=['writebw'])
df4 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-status/HM-Hybrid/hmhybrid-resilver.csv', names=['writebw'])
#df5 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-wb/writebw-hmls.csv', names=['writebw'])
df5 = pd.read_csv(f'/home/surbhi/measurements/zfs-resilvering-status/HM-LS/hmls-resilver.csv', names=['writebw'])


# 90% zone utilization
#df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-wb/writebw-cmr.csv')
df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-wb/writebw-cmr.csv')
data_sorted = np.sort(df['duration'])
cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
plt.plot(data_sorted, cdf, linestyle='-', 
         label="CMR", color=colors['90%'], 
         linewidth=2.5, alpha=0.85)

# 75% zone utilization
df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-status/DM-SMR/dm-smr-resilvering-bw.csv')
data_sorted = np.sort(df['duration'])
cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
plt.plot(data_sorted, cdf, linestyle='--', 
         label="DM-SMR", color=colors['75%'], 
         linewidth=2.5, alpha=0.85)

# 50% zone utilization
df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-status/HA-SMR/ha-smr-resilver-bw.csv')
data_sorted = np.sort(df['duration'])
cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
plt.plot(data_sorted, cdf, linestyle='-.', 
         label="HA-SMR", color=colors['50%'], 
         linewidth=2.5, alpha=0.85)

# 25% zone utilization
df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-wb/writebw-hmhybrid.csv')   
data_sorted = np.sort(df['duration'])
cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
plt.plot(data_sorted, cdf, linestyle=':', label="HM-Hybrid", color=colors['25%'], linewidth=3.0, alpha=0.85)


df = convert_bandwidth_to_latency_df('/home/surbhi/measurements/zfs-resilvering-status/HM-LS/hmls-resilver.csv')   
data_sorted = np.sort(df['duration'])
cdf = np.arange(1, len(data_sorted) + 1) / len(data_sorted)
plt.plot(data_sorted, cdf, linestyle=':', 
         label="HM-LS", color=colors['extra'], 
         linewidth=3.0, alpha=0.85)

# Set log scale for x-axis
plt.xscale('log', base=2)

# Clean styling - no grid
plt.grid(False)

# Add percentile reference lines
percentiles = [0.25, 0.5, 0.75, 0.9, 0.95, 0.99]  # 25th, 50th, 75th, 90th, 95th, 99th percentiles
for p in percentiles:
    plt.axhline(y=p, color='gray', linestyle='--', alpha=0.4, linewidth=1, zorder=0)

# Set ticks and labels with better spacing - include key percentiles
ytick_positions = [0, 0.25, 0.5, 0.75, 0.9, 0.95, 0.99]
ytick_labels = ['0%', '25%', '50%', '75%', '90%', '95%', '99%']
plt.yticks(ytick_positions, ytick_labels, fontsize=10)
plt.xticks(fontsize=11)

# Labels and title with improved typography
plt.xlabel('Latency (s)', fontsize=12, fontweight='medium')
plt.ylabel('CDF', fontsize=12, fontweight='medium')
plt.title('DMHybrid Performance: 1MB Zipf (90/10) Pattern', 
          fontsize=13, fontweight='semibold', pad=15)

# Improved legend
plt.legend(fontsize=10, frameon=True, fancybox=False, 
           shadow=False, framealpha=0.9, edgecolor='lightgray',
           loc='lower right')

# Reduce whitespace around the plot
plt.tight_layout(pad=0.5)

# Remove top and right spines for cleaner look
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

# Uncomment to save with high quality
plt.savefig("ZFS_resilvering_CDF.png", bbox_inches='tight', dpi=300, facecolor='white')

plt.show()
```

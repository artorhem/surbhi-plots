```python
import matplotlib.pyplot as plt
#import numpy as np
import pandas as pd
import math


names_latency=['time_msec', 'latency', 'op2', 'write_size', 'op3']
```

```python

df_90_lba = pd.read_csv('/home/surbhi/measurements/worst_case/lsdm/90Util/1MB/8TB/90-10-LBA/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_90_lba['time_msec'] = df_90_lba['time_msec'] - df_90_lba.iloc[0].time_msec
df_90_lba['time_seconds'] = df_90_lba['time_msec']/1000

# Round down the time_seconds to the nearest second
df_90_lba['time_seconds'] = df_90_lba['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_90_lba_ls = df_90_lba.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_90_lba_ls.columns = ['time_seconds', 'total_writes']
new_df_90_lba_ls['total_writes'] = new_df_90_lba['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_90_lba_ls.shape)

##################################################


df_75_lba = pd.read_csv('/home/surbhi/measurements/worst_case/lsdm/75Util/1MB/8TB/90-10-LBA/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_75_lba['time_seconds'] = df_75_lba['time_msec']/1000

# Round down the time_seconds to the nearest second
df_75_lba['time_seconds'] = df_75_lba['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_75_lba_ls = df_75_lba.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_75_lba_ls.columns = ['time_seconds', 'total_writes']
new_df_75_lba_ls['total_writes'] = new_df_75_lba['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_75_lba_ls.shape)

##################################################

df_50_lba = pd.read_csv('/home/surbhi/measurements/worst_case/lsdm/50Util/1MB/8TB/90-10-LBA/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_50_lba['time_seconds'] = df_50_lba['time_msec']/1000

# Round down the time_seconds to the nearest second
df_50_lba['time_seconds'] = df_50_lba['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_50_lba_ls = df_50_lba.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_50_lba_ls.columns = ['time_seconds', 'total_writes']
new_df_50_lba_ls['total_writes'] = new_df_50_lba_ls['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_75_lba.shape)

##################################################

df_99_uni = pd.read_csv('/home/surbhi/measurements/worst_case/STL/1M/uniform-random/run2/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_99_uni['time_seconds'] = df_99_uni['time_msec']/1000

# Round down the time_seconds to the nearest second
df_99_uni['time_seconds'] = df_99_uni['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_99_uni = df_99_uni.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_99_uni.columns = ['time_seconds', 'total_writes']
new_df_99_uni['total_writes'] = new_df_99_uni['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_99_uni.shape)

#######
df_99_zoned = pd.read_csv('/home/surbhi/measurements/worst_case/STL/1M/90-10-zoned/run2/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_99_zoned['time_seconds'] = df_99_zoned ['time_msec']/1000

# Round down the time_seconds to the nearest second
df_99_zoned ['time_seconds'] = df_99_zoned ['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_99_zoned  = df_99_zoned.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_99_zoned.columns = ['time_seconds', 'total_writes']
new_df_99_zoned['total_writes'] = new_df_99_zoned['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_99_zoned.shape)

#######
df_99_lba = pd.read_csv('/home/surbhi/measurements/worst_case/STL/1M/90-10-LBA/50Util/run2/lat_log_avg_lat.1.log', names=names_latency)
#process the bw data
# Convert 'time_msec' to seconds
df_99_lba['time_seconds'] = df_99_lba['time_msec']/1000

# Round down the time_seconds to the nearest second
df_99_lba['time_seconds'] = df_99_lba['time_seconds'].astype(int)

# Create a new dataframe with the count of writes per second
new_df_99_lba = df_99_lba.groupby('time_seconds')['write_size'].sum().reset_index()

# Rename columns for clarity
new_df_99_lba.columns = ['time_seconds', 'total_writes']
new_df_99_lba['total_writes'] = new_df_99_lba['total_writes'].cumsum() / 1024**3  # Convert bytes to MB

# Display the new dataframe
print(new_df_99_lba.shape)



```

```python
import matplotlib.pyplot as plt
import numpy as np

# Get the Paired colormap
cmap = plt.get_cmap('Paired')

# Convert RGB to Grayscale using luminance formula
def rgb_to_grayscale(rgb):
    # The input `rgb` is a tuple of (R, G, B, A), we only want the first three (R, G, B)
    return 0.2989 * rgb[0] + 0.5870 * rgb[1] + 0.1140 * rgb[2]

# Generate a grayscale version of the Paired colormap
grayscale_colors = [rgb_to_grayscale(cmap(i)) for i in range(cmap.N)]

# Sample line plot using the grayscale colors
x = np.linspace(0, 10, 100)
fig, ax = plt.subplots()

for i, gray in enumerate(grayscale_colors[:-1]):  # Plot using first 5 grayscale colors
    ax.plot(x, np.sin(x + i), color=str(gray), label=f'Line {i+1}')

ax.legend()
plt.show()

#now plot all colors in the colormap
fig1, ax1 = plt.subplots()
for i in [10, 8, 3]:
    ax1.plot(x, np.sin(x + i), color=cmap(i), label=f'Line {i+1}')

ax1.legend()
plt.show()

10, 8, 3

```
